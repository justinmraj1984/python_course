# Python Data Structures

## 1. List

```python
family_members = ["justin","jenifer","patrick","arthur"]
print(family_members)
```

```python
marks = [1,2,3,4]
type(marks)
```

```python
# Heterogenous collection of data
family_member_ages = ["justin",42,"jenifer",37,"patrick",12,"arthur",6]
print(family_member_ages)
```

```python
fruits_list = ["apple","grapes","pomegranete","mango","chickoo"]
print(fruits_list[2])
```

```python
# list allows duplicates
animal_list = ["cat", "dog", "rabbit"]
animal_list.append("dog")
print(animal_list)
```

```python
# List element traversal from left to right
print(fruits_list[2:4])
```

```python
# List element traversal from right to left - reverse order
print(fruits_list[-3:-1])
```

```python
sample_str = "Hello World !"
reverse_str = sample_str[::-1]
print(reverse_str)
```

```python
# palindrome program
while True:
    input_str = input("Enter string: ")
    if input_str == input_str[::-1]:
        print("Input string is Palyndrome !")
        break
    else:
        print("Input string is not a Palyndrome !")
```

## 2. Set

```python
new_set=set() # empty set
new_set.add("element1")
new_set.add("element2")
print(new_set)
```

```python
fruits_set={'mango', 'grapes', 'apple','drumstick'}
fruits_set.add("chickoo")
fruits_set.add("pomegranate")
fruits_set.discard("drumstick")
fruits_set.add("pomegranate")
print(fruits_set)
type(fruits_set)
```

```python
# iterate the elements in a set
data_set={"data", 100, 12.5}
for i in data_set:
    print(type(i), " : ", i)
```

```python
# convert a set to list
items_to_buy={'groceries','vegetables','stationary items'}
shopping_list = list(items_to_buy)

print(type(items_to_buy))
print(type(shopping_list))
```
