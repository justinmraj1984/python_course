### #2 - String Methods in Python

#### 1. capitalize()
Converts the first character to uppercase and the rest to lowercase.

```python
text = "hello WORLD"
print(text.capitalize())  # Output: "Hello world"

username = "john"
print(username.capitalize())  # Output: "John"
```

#### 2. casefold()
Converts the string to lowercase for aggressive, caseless matching.

```python
german_word = "Fußball"
print(german_word.casefold())  # Output: "fussball"

user_input = "Admin"
print(user_input.casefold() == "admin")  # Output: True
```

#### 3. center(width, fillchar)
Centers the string within a specified width using a padding character.

```python
title = "Menu"
print(title.center(10, "="))  # Output: "===Menu==="

code = "42"
print(code.center(6, "0"))  # Output: "004200"
```

#### 4. count(sub, start, end)
Returns the number of occurrences of a substring.

```python
text = "banana"
print(text.count("a"))  # Output: 3

phrase = "one apple, two apples"
print(phrase.count("apple"))  # Output: 2
```

#### 5. encode(encoding, errors)
Returns an encoded byte version of the string.

```python
text = "pythön"
print(text.encode())  # Output: b'pyth\xc3\xb6n'

symbol = "€"
print(symbol.encode("utf-8"))  # Output: b'\xe2\x82\xac'
```

#### 6. endswith(suffix)
Returns `True` if the string ends with the specified suffix.

```python
filename = "document.pdf"
print(filename.endswith(".pdf"))  # Output: True

url = "https://example.com"
print(url.endswith(".org"))  # Output: False
```

#### 7. expandtabs(tabsize)
Sets the tab size of the string in spaces.

```python
text = "a\tb\tc"
print(text.expandtabs(2))  # Output: "a b c"

data = "Name:\tAlice"
print(data.expandtabs(4))  # Output: "Name:   Alice"
```

#### 8. find(sub)
Returns the lowest index of the substring (-1 if not found).

```python
phrase = "find the needle"
print(phrase.find("needle"))  # Output: 9

print(phrase.find("haystack"))  # Output: -1
```

#### 9. format(*args, **kwargs)
Formats specified values and inserts them into placeholders `{}`.

```python
template = "Hello, {}!"
print(template.format("Alice"))  # Output: "Hello, Alice!"

item = "Item: {name}, Price: ${price}"
print(item.format(name="Book", price=12))  # Output: "Item: Book, Price: 12"
```

#### 10. format_map(mapping)
Formats values using a dictionary or mapping object.

```python
data = {"x": 10, "y": 20}
print("Point: ({x}, {y})".format_map(data))  # Output: "Point: (10, 20)"

user = {"name": "Bob"}
print("User: {name}".format_map(user))  # Output: "User: Bob"
```

### 5. User Defined Functions without For Loop, If Statement

```python
# Example 1: Calculate rectangle area
def calculate_area(width, height):
    return width * height

print(calculate_area(5, 4))

# Example 2: Convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print(celsius_to_fahrenheit(25))
```

### 6. User Defined Functions with For Loop

```python
# Example 1: Sum elements in a list
def sum_list(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print("sum: ",sum_list([10, 20, 30]))

# Example 2: Print a character multiple times across lines
def repeat_symbol(symbol, times):
    for _ in range(times):
        print(symbol)

repeat_symbol("*", 3)
```

### 7. User Defined Functions with For Loop and If Condition

```python
# Example 1: Count even numbers in a list
def count_evens(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count

print(count_evens([1, 2, 3, 4, 6]))

# Example 2: Find all names starting with a specific letter
def filter_names(names, letter):
    matches = []
    for name in names:
        if name.startswith(letter):
            matches.append(name)
    return matches

print(filter_names(["Alice", "Bob", "Anna"], "A"))
```

## Task 3 - Rock, Paper, Scissors Game

#### Build a simple Rock-Paper-Scissors game where the player competes with the computer.
This task will test your understanding of:
- Input/Output
- Conditional logic
- Random module
- Loops
- Exception handling

```python
import random

def get_user_choice():
    print("Select your Choice : [R]ock, [P]aper, [S]cissors or [Q]uit")
    user_choice = input(">")
    print("")

    if user_choice.lower() not in ["r","p","s","q"]:
        raise ValueError("Invalid Choice! Please enter the options - [R]ock, [P]aper, [S]cissors or [Q]uit")
        
    return user_choice.lower()

def play_game():
    round_num = 0
    user_score = 0
    computer_score = 0
    user_choice = ""
    choices = {"r":"Rock", "p":"Paper", "s":"Scissors"}
    
    print("***********************************************")
    print("*** Welcome to Rock - Paper - Scissors game ***")
    print("***********************************************")
    print("")
    user_name = input("> Enter your Name: ")    

    while True:
        round_num = round_num + 1
        print("")
        print(f"Round #{round_num}")
        
        computer_choice = random.choice(["r","p","s"])
        
        try:
            user_choice = get_user_choice()
        except ValueError as error:
            print(" * Error: " + str(error))
            computer_score = computer_score + 1
            print(" > Outcome: Computer Wins !")
            continue

        if user_choice.lower() == 'q':
            if user_score + computer_score == 0:
                break
            else:
                print("*********************************************************")
                print("***                    R E S U L T                    ***")
                print("*********************************************************")
                print(f"  Total Rounds : {round_num-1}")

                if round_num-1 != user_score + computer_score:
                    print(f"  Tie Rounds   : {round_num - user_score - computer_score - 1}")
                
                print(f"  Final Score --------------------------------")
                print(f"   - {user_name} : {user_score}")
                print(f"   - Computer : {computer_score}")
                if user_score > computer_score:
                    print(f"  Result: Congrats {user_name}, You Win !")
                elif computer_score > user_score:
                    print("  Result: Computer Wins !")
                else:
                    print("  Result: Its a Tie !")
                print("*********************************************************")
                break

        input_choices = user_choice + computer_choice
        print(f" > Your Choice: {choices[user_choice]}")
        print(f" > Computer Choice: {choices[computer_choice]}")

        # Winning Combination:
        # R - S : Rock crushes Scissor
        # P - R : Paper covers Rock
        # S - P : Scissor cuts Paper

        if input_choices in ["rs","pr","sp"]:
            print(" > Outcome: You win this Round")
            user_score = user_score + 1
        else:
            if user_choice == computer_choice:
                print(" > Outcome: This Round is a Tie !")
            else:
                print(" > Outcome: Computer wins this Round")
                computer_score = computer_score + 1
```

```python
play_game()
```

## Task 4 - Word Guessing Game (Mini Hangman)

#### Create a word-guessing game where the player tries to guess the letters of a hidden word.
This will test your:
- Loops
- Conditional logic
- String manipulation
- Exception handling
- Logical thinking

```python
import random

words_list = ["python", "machine", "learning", "bridge", "computer", "science", "data", "analysis", "platform", "visual"]
no_of_attempts = 10

def find_char_occurences(word, x):
    indices = []
    for char in range(len(word)):
        if word[char] == x:
            indices.append(char)
    return indices

def generate_display_word(word, count, indices, replacement_char):
    disp_word = word
    
    for x in range(count):
        disp_word = disp_word[:indices[x]] + replacement_char + disp_word[indices[x]+1:]
    return disp_word  

print("".center(50, "*"))
print("Welcome to the Word Guessing game !".center(50, " "))
print("".center(50, "*"))
print("Instructions :")
print("- Enter Only one alphabet character per attempt")
print("- Please avoid Numbers and Special Characters.")
print(f"- You have totally {no_of_attempts} attempts.")
print("".center(50, "*"))
print("")

# Select the Random word
random_word = random.choice(words_list)

# Fill the displayed clue with dashes
display_word = "".center(len(random_word), "-")
print(f"Guess the below word")
print(f"' {display_word} '")

for i in range(1, no_of_attempts+1):
    print("")
    user_input = input(f"Attempt #{i} > ")

    # Validate user input
    if user_input.lower().isalpha() and len(user_input)==1:
        user_input = user_input.lower()
        pass
    else:
        print("  * Error: Invalid Input, Read the instructions above and Try again..")
        continue
    
    # 1. Check if the user_input is alredy used
    char_exists = display_word.count(user_input)
    if char_exists > 0:
        print("  * Oops, Already used it..")
        continue

    # 2. Check if the user_input exists in the random_word, and no. of occurences
    char_count = random_word.count(user_input)
    if char_count == 0:
        print("  * Oops, Wrong Guess.. try again !")
        continue

    char_index = find_char_occurences(random_word, user_input)
    
    # 3. If exists, then display the characters in that position alone
    display_word = generate_display_word(display_word, char_count, char_index, user_input)
    print("")
    print(f"' {display_word} '")
    
    if display_word != random_word:
        print(" > Correct Guess, keep going !")

    # 4. Check if the random_word is fully identified and complete the game
    if display_word == random_word:
        print("")
        print("".center(50, "*"))
        print(f"Congrats ! You found the word in {i} attempts".center(50, " "))
        print("".center(50, "*"))
        break

# 5. All attempts are over and the Word is not guessed
if display_word != random_word:
    print("")
    print("".center(50, "*"))
    print(f"Game Over, no more attempts !".center(50, " "))
    print("")
    print(f" Correct word is ' {random_word} ', Better luck next time !")
    print("".center(50, "*"))    
```
