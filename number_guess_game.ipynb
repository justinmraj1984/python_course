# Number Guessing Game

## Level 1 - IF condition

```python
number_to_be_found = 86
```

```python
user_input = input("Guess the number: ")
```

```python
if number_to_be_found == user_input:
    print("Correct !")
else:
    print("InCorrect !")
```

## Level 2.1 - For Loop

```python
for i in range(0, 10):
    user_input = int(input("Guess the number: "))
    
    if number_to_be_found == user_input:
        print("Correct !")
        break
    else:
        print(number_to_be_found, " : ", user_input)
        print("InCorrect - try again !")
```

## Level 2.2 - While Loop

```python
while True:
    user_input = int(input("Guess the number: "))
    
    if number_to_be_found == user_input:
        print("Correct !")
        break
    else:
        # print(number_to_be_found, " : ", user_input)
        print("InCorrect - try again !")
```
