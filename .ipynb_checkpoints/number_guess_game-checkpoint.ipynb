## Level 3 - Randomise Secret Number and Validate user input

1. Generate the secret number randomly between 1 and 100
2. User guess should be between 1 and 100 else return 'Invalid Number'
3. No. of attempts to Guess the secret number is 10
5. Change the incorrect message in reference to the Secret number
6. Display the no. of attempts to guess the secret number correctly

```python
import random

secret_number = random.randint(1, 100)

print("**********************************************")
print("*~~  Welcome to the Number Guessing Game   ~~*")
print("*~ You have 10 attempts to Guess the number ~*")
print("**********************************************")

for i in range(1, 10):
    print("")
    user_input = int(input(f"Attempt #{i} - Guess the Secret Number: "))

    if user_input == secret_number:
        print(f"> Congratulations, You have guessed the Secret Number in {i} attempts !")
        break

    else:
        if user_input < 1 or user_input > 100:
            print(f"> Your input {user_input} is Invalid !")
            print("> Please guess a number between 1 and 100..")

        elif user_input < secret_number:
            print(f"> Your input {user_input} is less than the Secret Number !")

        elif user_input > secret_number:
            print(f"> Your input {user_input} is greater than the Secret Number !")

    if i < 10:
        print(f"> {10-i} attempts left..")
        print("> Please try again..")
```

## Level 4 - Add Exception Handler for ValueError

1. ValueError occurs at runtime when the user enters a non-numerical value as input

```python
import random
secret_number = random.randint(1, 100)

print("**********************************************")
print("*~~  Welcome to the Number Guessing Game   ~~*")
print("*~ You have 10 attempts to Guess the number ~*")
print("**********************************************")

for i in range(1, 11):
    print("")
    try:
        user_input = int(input(f"Attempt #{i} - Guess the Secret Number: "))
    except ValueError:
        print(f"> ERROR: Your input is not a valid Integer !")
        print(f"> {10-i} more attempts left..")
        continue

    if user_input == secret_number:
        print(f"> Congratulations, You have guessed the Secret Number in {i} attempts !")
        break

    else:
        if user_input < 1 or user_input > 100:
            print(f"> Your input {user_input} is Invalid !")
            print("> Please guess a number between 1 and 100..")

        elif user_input < secret_number:
            print(f"> Your input {user_input} is less than the Secret Number !")

        elif user_input > secret_number:
            print(f"> Your input {user_input} is greater than the Secret Number !")

    if i < 10:
        print(f"> {10-i} more attempts left..")
        print("> Please try again..")
    elif i == 10:
        print("> No more attempts left..")
```
